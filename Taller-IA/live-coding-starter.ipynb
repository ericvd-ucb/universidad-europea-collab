## 5. A loop

```python
for t in temps:
    print(t, "degrees")
```

## 6. A decision

```python
for t in temps:
    if t > 20:
        print(t, "is warm")
    else:
        print(t, "is cool")
```

## 7. Your own function

The formula in Markdown:

$$ A = P(1 + r)^t $$

The same formula in Python:

```python
def grow(amount, rate, years):
    return amount * (1 + rate) ** years

grow(1000, 0.05, 10)
```

## 9. A table of data

```python
import pandas as pd

df = pd.DataFrame({
    "month": ["Jan", "Feb", "Mar", "Apr", "May"],
    "temp": temps
})
df
```

## 10. A plot

```python
import matplotlib.pyplot as plt

plt.plot(df["month"], df["temp"], marker="o")
plt.title("Monthly temperature")
plt.ylabel("Degrees C")
plt.show()
```
